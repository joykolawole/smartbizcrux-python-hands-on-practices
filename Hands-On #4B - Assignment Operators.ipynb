```python
## Inventory Management System (Mini Project)

stock = 150
stock -= 28
stock += 75
stock //= 12
print (stock)

#🌟 Bonus Challenge

stock = int(input("enter the initial stock:"))
sold = int(input("enter items sold:"))
new_arrival = int(input("enter new items arrived:"))
carton = int(input("enter how many items fit into one carton:"))

stock -= sold
print("Remaining stock after stock:", stock)

stock += new_arrival
print("Remaining stock after sales", stock)

stock //= carton
print("Number of full cartons", stock)
stock %= carton
print("Number of remainder", stock)
```
