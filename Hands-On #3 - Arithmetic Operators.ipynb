```python
# Smart Shopping Discount Calculator

laptop_bag = 18500
wireless_mouse = 12000
keyboard = 24500
USB_drive = 5000
print("Total Purchase:", (laptop_bag + wireless_mouse + keyboard + USB_drive))
print("Discount Qualified:", (laptop_bag + wireless_mouse + keyboard + USB_drive) >= 50000)
print("Discount Amount:", (laptop_bag + wireless_mouse + keyboard + USB_drive) // 10)
print("Final Amount:", (laptop_bag + wireless_mouse + keyboard + USB_drive) - ((laptop_bag + wireless_mouse + keyboard + USB_drive) // 10))
```
